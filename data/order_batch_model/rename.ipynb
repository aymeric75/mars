```python
import os

old_part_1 = "LOBSTER_LOBSTER-"
old_part_2 = "_order_images"

for name in os.listdir("."):
    if name.endswith(".zarr"):
        new_name = name.replace(old_part_1, "").replace(old_part_2, "")

        if new_name != name:
            print(f"Renaming:\n  {name}\n→ {new_name}")
            os.rename(name, new_name)
```
